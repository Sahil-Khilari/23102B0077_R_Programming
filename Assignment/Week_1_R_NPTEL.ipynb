## 5. Command line, scripts and comments

```r
# In R, # starts a comment. Assignment commonly uses <-.
student_name <- "Student"
marks <- c(82, 76, 91, 88, 79)
mean_marks <- mean(marks)
mean_marks

# A small reusable sequence of commands (like a script).
result <- data.frame(
  Student = student_name,
  Marks = marks,
  Passed = marks >= 40
)
print(result)
```

## 6. Data-editor concept in a browser notebook

```r
# R's interactive data.entry() / fix() editors depend on a desktop GUI
# and may not work in Colab. A data frame can still be created and modified in code.
students <- data.frame(
  Name = c("Asha", "Ravi", "Neha"),
  Marks = c(78, 85, 92)
)
print(students)

# Update a value programmatically
students$Marks[students$Name == "Ravi"] <- 89
print(students)
```
