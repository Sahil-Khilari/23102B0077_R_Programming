## 2. Comments, vectors, mode and storage mode

```python
%%R
# The c() function combines values into a vector.
numbers <- c(1, 2, 3, 4, 5)
characters <- c("red", "green", "blue")
logical_values <- c(TRUE, FALSE, TRUE)

numbers
mode(numbers)
storage.mode(numbers)
mode(characters)
storage.mode(characters)
mode(logical_values)
storage.mode(logical_values)

# R vectors generally coerce elements to a common type.
mixed <- c(1, 2, "three")
mixed
str(mixed)
```

## 4. Arithmetic with data vectors

```python
%%R
a <- c(2, 3, 5, 7)
b <- c(-2, -3, -5, 8)

a + b
a - b
a * b
a / c(1, 1, 5, 2)

# R recycles a shorter vector when lengths are compatible.
c(2, 3, 5, 7) + c(8, 9)

# A non-multiple length produces a warning about recycling.
# Run separately to observe the warning:
c(2, 3, 5, 7) + c(8, 9, 10)
```

## 5. Infinity and finite values

```python
%%R
positive_infinity <- 3 / 0
negative_infinity <- -3 / 0
undefined_result <- 0 / 0

positive_infinity
negative_infinity
undefined_result

is.finite(c(1, positive_infinity, undefined_result))
is.infinite(c(1, positive_infinity, negative_infinity))
is.nan(undefined_result)
is.na(undefined_result)
```

### Key observation
R performs vectorized arithmetic element by element. If vectors have different lengths, the shorter one may be recycled; if the longer length is not a multiple of the shorter length, R warns. The notebook demonstrates this behavior as shown in the Week 2 notes.
